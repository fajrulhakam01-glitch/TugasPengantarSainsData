```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

print("Dataset berhasil dibaca!")
```

```python
df.info()
```

```python
df.head()
```

```python
missing_values = df.isnull().sum()

print("Jumlah Missing Value:")
print(missing_values)
```

```python
print("Kolom yang memiliki Missing Value:")
print(missing_values[missing_values > 0])
```

```python
df['kode_kasir'] = df['kode_kasir'].fillna('TIDAK DIKETAHUI')

print("Missing Value Setelah Penanganan:")
print(df.isnull().sum())
```

```python
df['harga_satuan'] = pd.to_numeric(
    df['harga_satuan'].astype(str)
    .str.replace(',00', '', regex=False)
    .str.replace('.', '', regex=False),
    errors='coerce'
)
```

```python
df['tanggal'] = pd.to_datetime(
    df['tanggal'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
```

```python
df.info()
```

```python
print("Jumlah Transaksi Berdasarkan Status:")
print(df['status'].value_counts())
```

```python
df_valid = df[df['status'] == 'SELESAI'].copy()

print("Jumlah Transaksi Berhasil:", len(df_valid))
```

```python
# Menghitung harga setelah diskon
df_valid['harga_setelah_diskon'] = (
    df_valid['harga_satuan'] *
    (1 - df_valid['diskon_persen'] / 100)
)

# Menghitung total pendapatan setiap transaksi
df_valid['pendapatan_bersih'] = (
    df_valid['harga_setelah_diskon'] *
    df_valid['jumlah']
)

df_valid.head()
```

```python
df_valid['tahun'] = df_valid['tanggal'].dt.year

df_valid[['tanggal', 'tahun', 'jumlah', 'harga_satuan',
          'diskon_persen', 'pendapatan_bersih']].head()
```

```python
pendapatan_tahunan = (
    df_valid[df_valid['tahun'].isin([2024, 2025])]
    .groupby('tahun')['pendapatan_bersih']
    .sum()
    .reset_index()
)

pendapatan_tahunan.columns = ['Tahun', 'Total Pendapatan Bersih']

pendapatan_tahunan
```

```python
hasil = pendapatan_tahunan.copy()

hasil['Total Pendapatan Bersih'] = hasil[
    'Total Pendapatan Bersih'
].apply(lambda x: f"Rp {x:,.0f}".replace(',', '.'))

print("TOTAL PENDAPATAN BERSIH PER TAHUN")
display(hasil)
```

```python
df_valid.to_csv('hasil_analisis_transaksi.csv', index=False)

print("Hasil analisis berhasil disimpan!")
```
