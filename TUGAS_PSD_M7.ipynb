```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Cell ini digunakan untuk mengimpor library yang dibutuhkan dalam proses pengolahan dan visualisasi data. Library Pandas (pd) digunakan untuk membaca, mengolah, dan menganalisis dataset dalam bentuk tabel. NumPy (np) digunakan untuk mendukung operasi numerik atau perhitungan data. Sedangkan Matplotlib (plt) digunakan untuk membuat visualisasi data seperti grafik batang, diagram lingkaran, dan scatter plot.

```python
#membaca dataset
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')
df.head()
```

Cell ini digunakan untuk membaca dataset transaksi POS tahun 2024–2025 menggunakan fungsi pd.read_csv(). Data yang berhasil dibaca disimpan ke dalam variabel df agar dapat diolah lebih lanjut. Fungsi df.head() digunakan untuk menampilkan lima baris pertama dari dataset sehingga kita dapat melihat struktur dan isi data, seperti tanggal transaksi, kode produk, jumlah pembelian, harga satuan, diskon, dan status transaksi.

```python
#mengubah format tanggal
df['tanggal'] = pd.to_datetime(
    df['tanggal'],
    format='mixed',
    dayfirst=True
)

#mengubah harga satuan menjadi numerik
df['harga_satuan'] = pd.to_numeric(
    df['harga_satuan'].astype(str).str.replace(',', '.', regex=False)
)

#mengambil data tahun 2025
df_2025 = df[df['tanggal'].dt.year == 2025].copy()

#menghitung total penjualan setelah diskon
df_2025['total_penjualan'] = (
    df_2025['jumlah'] *
    df_2025['harga_satuan'] *
    (1 - df_2025['diskon_persen'] / 100)
)

#mengambil transaksi yang berhasil
df_selesai = df_2025[df_2025['status'] == 'SELESAI']

print("Jumlah data transaksi tahun 2025:", len(df_2025))
df_2025.head()
```

Cell ini digunakan untuk membersihkan dan mengolah data sebelum dilakukan analisis. Pertama, format tanggal diubah menggunakan pd.to_datetime() agar dapat dikenali sebagai tipe data tanggal. Selanjutnya, kolom harga satuan diubah menjadi numerik menggunakan pd.to_numeric() agar dapat digunakan dalam perhitungan.

Setelah itu, data difilter untuk mengambil transaksi yang terjadi pada tahun 2025. Kemudian, dibuat kolom baru bernama total_penjualan untuk menghitung total penjualan berdasarkan jumlah barang, harga satuan, dan persentase diskon. Terakhir, data transaksi dengan status SELESAI dipisahkan ke dalam variabel df_selesai agar dapat digunakan dalam analisis penjualan.

```python
fig, ax = plt.subplots(2, 2, figsize=(18, 12))
```

Cell ini digunakan untuk membuat area visualisasi yang terdiri dari 2 baris dan 2 kolom menggunakan Matplotlib. Ukuran keseluruhan area grafik ditentukan sebesar 18 × 12 inci. Variabel fig digunakan untuk menyimpan keseluruhan area gambar, sedangkan ax digunakan untuk mengakses masing-masing area grafik.

```python
#Menghitung total penjualan setiap bulan
penjualan_bulanan = df_selesai.groupby(
    df_selesai['tanggal'].dt.month
)['total_penjualan'].sum()

#Nama bulan
nama_bulan = [
    'Januari', 'Februari', 'Maret', 'April',
    'Mei', 'Juni', 'Juli', 'Agustus',
    'September', 'Oktober', 'November', 'Desember'
]

#Melengkapi bulan yang tidak memiliki transaksi
penjualan_bulanan = penjualan_bulanan.reindex(
    range(1, 13),
    fill_value=0
)

#Membuat grafik
plt.figure(figsize=(12, 6))
plt.bar(nama_bulan, penjualan_bulanan.values)

plt.title('Total Penjualan Bulanan Tahun 2025')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.tight_layout()

#Menyimpan grafik untuk dashboard
plt.savefig('/content/grafik1.png', dpi=300, bbox_inches='tight')

plt.show()
```

Cell ini digunakan untuk menghitung dan menampilkan total penjualan pada setiap bulan selama tahun 2025. Data dikelompokkan berdasarkan bulan menggunakan groupby(), kemudian total penjualan setiap bulan dijumlahkan menggunakan sum().

Selanjutnya, dibuat daftar nama bulan dari Januari hingga Desember. Fungsi reindex() digunakan untuk memastikan seluruh bulan tetap ditampilkan, termasuk bulan yang tidak memiliki transaksi. Setelah itu, grafik batang dibuat menggunakan plt.bar() untuk memperlihatkan perbandingan total penjualan setiap bulan. Grafik kemudian diberi judul, label sumbu, dan disimpan dalam format PNG agar dapat digunakan untuk membuat dashboard.

```python
produk_terlaris = df_selesai.groupby(
    'kode_produk'
)['jumlah'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.barh(produk_terlaris.index, produk_terlaris.values)

plt.title('10 Produk Terlaris Tahun 2025')
plt.xlabel('Jumlah Terjual')
plt.ylabel('Kode Produk')
plt.gca().invert_yaxis()
plt.tight_layout()

plt.savefig('/content/grafik2.png', dpi=300, bbox_inches='tight')

plt.show()
```

Cell ini digunakan untuk mengetahui 10 produk dengan jumlah penjualan terbanyak selama tahun 2025. Data dikelompokkan berdasarkan kode produk menggunakan groupby(), kemudian jumlah barang yang terjual dijumlahkan menggunakan sum().

Selanjutnya, data diurutkan dari jumlah penjualan terbesar ke terkecil menggunakan sort_values(). Fungsi head(10) digunakan untuk mengambil 10 produk teratas. Hasilnya ditampilkan dalam bentuk grafik batang horizontal menggunakan plt.barh() sehingga produk dengan penjualan tertinggi dapat terlihat dengan mudah.

```python
status_transaksi = df_2025['status'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    status_transaksi.values,
    labels=status_transaksi.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Persentase Status Transaksi Tahun 2025')
plt.tight_layout()

plt.savefig('/content/grafik3.png', dpi=300, bbox_inches='tight')

plt.show()
```

Cell ini digunakan untuk mengetahui perbandingan status transaksi selama tahun 2025, yaitu transaksi yang berhasil, dibatalkan, dan mengalami retur. Fungsi value_counts() digunakan untuk menghitung jumlah transaksi berdasarkan masing-masing status.

Selanjutnya, hasil perhitungan divisualisasikan menggunakan diagram lingkaran atau Pie Chart melalui plt.pie(). Parameter autopct digunakan untuk menampilkan persentase pada setiap bagian diagram, sedangkan startangle digunakan untuk menentukan posisi awal diagram. Visualisasi ini membantu mengetahui persentase keberhasilan dan status transaksi lainnya.

```python
# Mengambil data transaksi yang sudah selesai
data_diskon = df_selesai.copy()

# Menghitung korelasi antara diskon dan jumlah pembelian
korelasi = data_diskon[['jumlah', 'diskon_persen']].corr().iloc[0, 1]

# Membuat scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(
    data_diskon['diskon_persen'],
    data_diskon['jumlah'],
    alpha=0.5
)

plt.title('Hubungan Diskon dengan Jumlah Pembelian')
plt.xlabel('Diskon (%)')
plt.ylabel('Jumlah Pembelian')
plt.tight_layout()

# Menyimpan grafik untuk dashboard
plt.savefig('/content/grafik4.png', dpi=300, bbox_inches='tight')

plt.show()

print('Nilai korelasi:', korelasi)
```

Cell ini digunakan untuk mengetahui hubungan antara persentase diskon dengan jumlah barang yang dibeli oleh pelanggan. Data transaksi yang digunakan terlebih dahulu diambil dari transaksi yang sudah selesai menggunakan df_selesai.copy().

Selanjutnya, dilakukan perhitungan korelasi antara kolom jumlah pembelian dan diskon menggunakan fungsi corr(). Hasil korelasi digunakan untuk mengetahui arah dan kekuatan hubungan antara kedua variabel tersebut.

Kemudian, dibuat Scatter Plot menggunakan plt.scatter() untuk menampilkan hubungan antara persentase diskon pada sumbu X dan jumlah pembelian pada sumbu Y. Nilai korelasi juga ditampilkan menggunakan print() sebagai pendukung hasil analisis.

```python
from PIL import Image

#Membuka empat grafik yang sudah disimpan
gambar1 = Image.open('/content/grafik1.png').convert('RGB')
gambar2 = Image.open('/content/grafik2.png').convert('RGB')
gambar3 = Image.open('/content/grafik3.png').convert('RGB')
gambar4 = Image.open('/content/grafik4.png').convert('RGB')

#Menentukan ukuran setiap grafik
lebar = 800
tinggi = 500

gambar1 = gambar1.resize((lebar, tinggi))
gambar2 = gambar2.resize((lebar, tinggi))
gambar3 = gambar3.resize((lebar, tinggi))
gambar4 = gambar4.resize((lebar, tinggi))

#Membuat kanvas dashboard
dashboard = Image.new(
    'RGB',
    (lebar * 2, tinggi * 2),
    'white'
)

#Menempatkan grafik pada dashboard
dashboard.paste(gambar1, (0, 0))
dashboard.paste(gambar2, (lebar, 0))
dashboard.paste(gambar3, (0, tinggi))
dashboard.paste(gambar4, (lebar, tinggi))

#Menyimpan dan menampilkan dashboard
dashboard.save('/content/dashboard_visualisasi.png')

display(dashboard)
```

Cell ini digunakan untuk menggabungkan empat grafik yang telah dibuat sebelumnya menjadi satu tampilan dashboard. Pertama, library PIL diimpor untuk membantu membuka dan mengolah gambar. Selanjutnya, keempat grafik dibuka menggunakan Image.open() dan diubah ke format RGB.

Setiap gambar kemudian diatur ukurannya menjadi 800 × 500 piksel menggunakan resize(). Setelah itu, dibuat kanvas kosong berwarna putih menggunakan Image.new() dengan ukuran yang cukup untuk menampung empat grafik.

Keempat grafik ditempatkan pada posisi yang berbeda menggunakan paste(), yaitu dua grafik di bagian atas dan dua grafik di bagian bawah. Terakhir, dashboard disimpan dalam format PNG menggunakan save() dan ditampilkan menggunakan display().
